## TASEP projections

```python
# All MLQs projecting to a given state

state = [2, 4, 0, 1, 0]
lam = Partition([4,2,1])
n = 5
r = 0.2

mlq_shape = list(lam.conjugate())
show = False

aux_list = []
ball_arrangements = []

for i in mlq_shape:
    aux_list.append([list(sub) for sub in Subsets(range(n)) if len(sub) == i])

ball_arr_coords = itertools.product(*aux_list)

for tup in ball_arr_coords:
    ball_arr = []
    for row in tup:
        ball_row = [0 for j in range(n)]
        for i in range(n):
            if i in row:
                ball_row[i] = 1
        ball_arr.append(ball_row.copy())
    ball_arrangements.append(ball_arr)

all_good = True
cont = 0
list_mlqs = []
for bs in ball_arrangements:
    M = MultilineQueue(bs)
    M.pair(0)
    proj1 = M.projection()
    
    if proj1 == state:
        cont += 1
        list_mlqs.append(M)

print(len(list_mlqs))
print(len(ball_arrangements))
```

```python

# finds the set of ball arrangements that are reached from "balls" by a single transition in the Crystal Markov Chain
def going_out_from(balls,show):
    n = len(balls[0])
    M = MultilineQueue(balls)
    M.pair(0)
    proj1 = M.projection()
    rw0 = list(reversed(M.rw()))
    
    going_out = []

    if show:
        print("Going out:")
    
    for i in range(1,n+1):
        eps_i = M.epsilon(i)
        for j in range(1,eps_i+1):
            w = M.ring(i,j)
            if w != rw0 and w not in going_out and w != []:
                going_out.append(w)
                balls2 = ball_arrangement_from_word( w , M.shape, n)
                M2 = MultilineQueue(balls2)
                M2.pair(0)
                proj2 = M2.projection()
    
    if show:
        print(going_out)

    return(going_out)


# finds the set of ball arrangements that get to "balls" by a single transition in the Crystal Markov Chain
def going_in_to(balls,show):
    n = len(balls[0])
    M = MultilineQueue(balls)
    M.pair(0)
    rw0 = list(reversed(M.rw()))
    proj1 = M.projection()
    lam = M.partition()
    mlq_shape = list(lam.conjugate())
    n = M.size[0]

    aux_list = []
    ball_arrangements = []
    
    for i in mlq_shape:
        aux_list.append([list(sub) for sub in Subsets(range(n)) if len(sub) == i])
    
    ball_arr_coords = itertools.product(*aux_list)
    
    for tup in ball_arr_coords:
        ball_arr = []
        for row in tup:
            ball_row = [0 for j in range(n)]
            for i in range(n):
                if i in row:
                    ball_row[i] = 1
            ball_arr.append(ball_row.copy())
        ball_arrangements.append(ball_arr)

    going_in= []

    if show:
        print("Going in")
    
    for bs in ball_arrangements:    
        M2 = MultilineQueue(bs)
        M2.pair(0)
        proj2 = M2.projection()
        for i in range(1,n+1):
            eps_i = M2.epsilon(i)
            for j in range(1,eps_i+1):
                w = M2.ring(i,j)
                rw2 = list(reversed(M2.rw()))
                if w == rw0 and balls != bs and rw2 not in going_in:
                    going_in.append( rw2 )

    if show:
        print(going_in)
    
    return(going_in)
```

```python
# Checking condition for lumping

lam = Partition([4,3,1])
n = 4

mlq_shape = list(lam.conjugate())
show = False

aux_list = []
ball_arrangements = []

for i in mlq_shape:
    aux_list.append([list(sub) for sub in Subsets(range(n)) if len(sub) == i])

ball_arr_coords = itertools.product(*aux_list)

for tup in ball_arr_coords:
    ball_arr = []
    for row in tup:
        ball_row = [0 for j in range(n)]
        for i in range(n):
            if i in row:
                ball_row[i] = 1
        ball_arr.append(ball_row.copy())
    ball_arrangements.append(ball_arr)

all_good = True
for bs in ball_arrangements:
    M = MultilineQueue(bs)
    M.pair(0)
    proj1 = M.projection()

    # 1. If a_i < a_{i+1} then phi_i(M) > 0 and transition only happens for j = phi_i(M).

    for i in range(1,len(proj1)+1):
        if i < n:
            if proj1[i-1] < proj1[i]:
                phii = M.epsilon(i)
                if not phii > 0:
                    print("Mistake 1.")
                    all_good = False
                else:
                    for j in range(1,phii+1):
                        w = M.ring(i,j)
                        b2 = ball_arrangement_from_word(w,M.shape,M.size[0])
                        M2 = MultilineQueue(b2)
                        M2.pair(0)
                        proj2 = M2.projection()
                        
                        if j != phii and proj1 != proj2:
                            print("Mistake 2.")
                            all_good = False
                        elif j== phii and proj1 == proj2:
                            print("Mistake 3.")
                            all_good = False

        if i == n:
            if proj1[n-1] < proj1[0]:
                phii = M.epsilon(i)
                if not phii > 0:
                    print("Mistake 1.")
                    all_good = False
                else:
                    for j in range(1,phii+1):
                        w = M.ring(i,j)
                        b2 = ball_arrangement_from_word(w,M.shape,M.size[0])
                        M2 = MultilineQueue(b2)
                        M2.pair(0)
                        proj2 = M2.projection()
                        
                        if j != phii and proj1 != proj2:
                            print("Mistake 2.")
                            all_good = False
                        elif j== phii and proj1 == proj2:
                            print("Mistake 3.")
                            all_good = False

    #2 If epsilon_k(M) > 0 and transition only happens for j = epsilon_k(M) then a_k < a_{k+1}

    for k in range(1,n+1):
        epsk = M.epsilon(k)
        if epsk > 0:
            only_max = True
            for j in range(1,epsk+1):
                w = M.ring(k,j)
                b2 = ball_arrangement_from_word(w,M.shape,M.size[0])
                M2 = MultilineQueue(b2)
                M2.pair(0)
                proj2 = M2.projection()
                if j < epsk and proj1 != proj2:
                    only_max = False
                    print("Mistake 4.")
                    all_good = False
                    break

                if j == epsk and proj1 == proj2:
                    only_max = False
                    if k < n:
                        if proj1[k-1] < proj1[k]:
                            print("Mistake 5.")
                            all_good = False
                    elif k==n:
                        if proj1[n-1] < proj1[0]:
                            print("Mistake 5.")
                            all_good = False

            if only_max:
                if k < n:
                    if proj1[k-1] >= proj1[k]:
                        print("Mistake 6.")
                        all_good = False
                elif k==n:
                    if proj1[n-1] >= proj1[0]:
                        print("Mistake 6.")
                        all_good = False


if all_good:
    print("No mistakes!")      
```
